```python
import json
import math
import pandas as pd
def haversine (lat1, lon1, lat2, lon2):
    R = 6371  #
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (math.sin(dlat/2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon/2)**2)

    return 2 * R * math.asin(math.sqrt(a))


```

```python
lat1, lon1 = 28.5355, 77.3910
lat2, lon2 = 28.6139, 77.2090
distance = haversine(lat1, lon1, lat2,lon2)
print(f"Straight-line distance: {distance:.2f} km")
```

```python
from pyproj import Transformer
import math

def dms_to_decimal(deg, minute, sec, direction):
    value = abs(deg) + minute / 60 + sec / 3600
    if direction.upper() in ("S", "W"):
        value = -value
    return value

lat1 = dms_to_decimal(24, 36, 0.72, "N")
lon1 = dms_to_decimal(73, 40, 49.84, "E")

lat2 = dms_to_decimal(26, 55, 26, "N")
lon2 = dms_to_decimal(75, 49, 37, "E")

transformer = Transformer.from_crs(
    "EPSG:4979", "EPSG:4978", always_xy=True
)


p1 = transformer.transform(lon1, lat1, 0)
p2 = transformer.transform(lon2, lat2, 0)

print("Moti Magri ECEF (X, Y, Z):")
print(tuple(round(v, 3) for v in p1), "metres")

print("\nHawa Mahal ECEF (X, Y, Z):")
print(tuple(round(v, 3) for v in p2), "metres")

distance = math.dist(p1, p2)
print(f"\nECEF straight-line distance: {distance / 1000:.3f} km")
```
